## Podcasts summary test from markdown files

This notebook is for testing of summarization results of different prompts. The tested podcast is the [Data Intensive AI](https://datatalks.club/podcast/s20e05-data-intensive-ai.html).

The model used is `gpt-4o-mini`, which also can be changed later depeding on the performance and expectations. 

- **Note**: If you want to test it by yourself, you need to put the openAI api key to the `.env` file with the name `OPENAI_API_KEY`. 


```python
with open("data_intensive_ai_interview.md", "r", encoding="utf-8") as f:
    content = f.read()

print(content)
```

```python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()

OPENAI_API_KEY=os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=OPENAI_API_KEY)


model = "gpt-4o-mini"

def podcast_summary_generate(client, model, prompt):
    response = client.chat.completions.create(
    model=model,  
    messages=[
        {"role": "system", "content": "You are a helpful assistant that summarizes podcasts detailedly."},
        {"role": "user", "content": prompt}
    ],
    temperature=0.1,
    max_tokens=10000  # Adjust depending on how long you want the summary
)

# Extract and print the summary
    return response.choices[0].message.content

    
```

```python
prompt_1 = f'''You are an expert summarizer. Given a Markdown-formatted transcript of a podcast, your job is to summarize detailedly each chapter and extract important keywords.

Instructions:
- Identify chapters using the Markdown headings (e.g., ## Chapter 1: Topic).
- For each chapter:
  1. Provide a short paragraph summary written in third person, recapping what was discussed.
  2. List at most 5 relevant keywords that reflect the key topics or concepts of that chapter.
- Do not include dialogue or speaker names. 
- Use a clear and neutral tone.
- At the end create a 2 paragraph of summarization of whole topics without including the speaker names.
- Format your output like this:

## Chapter 1: [Title]
**Summary:** [your paragraph here]  
**Keywords:** keyword1, keyword2, keyword3, ...

## Chapter 2: [Title]
**Summary:** [your paragraph here]  
**Keywords:** keyword1, keyword2, ...

Here is the podcast transcript in Markdown:\n\n{content}.'''

summary_1 = podcast_summary_generate(client, model, prompt=prompt_1)
print("Summary:\n", summary_1)

```

```python
prompt_2 = f'''You are an expert summarizer for LinkedIn. Given a Markdown-formatted transcript of a podcast, your job is to create engaging, professional chapter summaries optimized for LinkedIn posts.

Instructions:
- Identify chapters using the Markdown headings.
- For each chapter:
  1. Add 1–2 relevant emojis **at the beginning of the chapter title** to make it visually engaging.
  2. Write a short paragraph summary (in third person) recapping the key discussion points.
  3. List up to 5 relevant keywords that reflect the core topics or concepts.
- Do not include dialogue or speaker names.
- Use a clear, neutral, and professional tone.
- At the end:
  - Write two summary paragraphs that encapsulate the entire podcast (without mentioning speaker names).
  - Add a final set of 3–5 emojis that reflect the overarching themes.

Here is the podcast transcript in Markdown:

{content}
'''

summary_2 = podcast_summary_generate(client, model, prompt=prompt_2)
print("Summary:\n", summary_2)
```
